# Robustness and Continuity

Evaluate explanation changes under a controlled input neighborhood. Continuity depends on the radius, norms, and neighborhood validity.

```python
import numpy as np
import matplotlib.pyplot as plt
from xai_eval.metrics import max_sensitivity

# A known synthetic explanation isolates the geometry of the metric.
def explanation(point):
    return np.array([point[0], 2 * point[1]])

x = np.array([0.0, 0.0])
radius = 0.1
angles = np.linspace(0, 2 * np.pi, 16, endpoint=False)
perturbations = radius * np.column_stack([np.cos(angles), np.sin(angles)])
value = max_sensitivity(explanation, x, perturbations, radius=radius)
print(f"Maximum attribution distance inside radius {radius}: {value:.3f}")
plt.scatter(perturbations[:, 0], perturbations[:, 1]); plt.scatter([0], [0], marker="x", color="black")
plt.gca().set_aspect("equal"); plt.title("Synthetic L2 neighborhood"); plt.show()
```
